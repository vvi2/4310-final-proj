```python
import requests
from bs4 import BeautifulSoup

import time
import json
```

```python
all_data = dict()
for i in range(2004, 2045):
    all_data.setdefault(i, dict())
    url = (
        "https://www.timeanddate.com/moon/phases/@5122440?year=" + str(i)
    )

    result = requests.get(url)

    # error handling
    if result.status_code != 200:
        print("something went wrong:", result.status_code, result.reason)

    # save html in case of errors
    with open("moon_data.html", "w", encoding="utf-8") as writer:
        writer.write(result.text)

    with open("moon_data.html", "r", encoding="utf-8") as reader:
        html_source = reader.read()

    page = BeautifulSoup(html_source, "html.parser")

    table_data = page.find_all("td", {"class": "tr"})
    counter = 1
    date = ""
    moon = ""
    moons = {1: "New Moon", 3: "First Quarter", 5: "Full Moon", 7: "Third Quarter"}
    # for j in moons.values():
    #     all_data[i].setdefault(j, [])
    for row in table_data:
        if ((counter == 1 or counter == 3 or counter == 5 or counter == 7) and row.text.strip() != ""):
            date = row.text.strip()
            moon = moons[counter]
        if ((counter == 2 or counter == 4 or counter == 6 or counter == 8) and date != "" and moon != ""):
            # all_data[i][moon].append(date + ", " + row.text.strip())
            # all_data[i][moon].append(date)
            all_data[i].setdefault(date, moon)

        if counter == 9:
            counter = 1
        else:
            counter += 1

    time.sleep(
        1
    )  # wait one second after each page is scraped to prevent overloading the website

print("done")
print(all_data)
```

```python
with open("reverse_moon_data.json", "w", encoding="utf-8") as writer:
  json.dump(all_data, writer, indent=2, ensure_ascii=False)
```
